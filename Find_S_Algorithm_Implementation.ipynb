```python
import pandas as pd
```

```python
data= {
    "Sky": ["Sunny", "Sunny", "Rainy", "Sunny"],
    "AirTemp": ["Warm", "Warm", "Cold", "Warm"],
    "Humidity": ["Normal", "High", "High", "High"],
    "Wind": ["Strong", "Weak", "Strong", "Strong"],
    "Water": ["Warm", "Warm", "Warm", "Cool"],
    "Forecast": ["Same", "Same", "Change", "Change"],
    "EnjoySport": ["Yes", "Yes", "No", "Yes"]
}

df= pd.DataFrame(data)
print(df)
```

```python
features= df.drop("EnjoySport", axis=1)
target= df["EnjoySport"]

print("Features:")
print(features)

print("\nTarget:")
print(target)
```

```python
hypothesis= ["Ø"] * len(features.columns)

print("Initial hypothesis:")
print(hypothesis)
```

```python
for i in range(len(features)):

    if target[i]=="Yes":
        example= features.iloc[i].values

        for j in range(len(hypothesis)):

            if hypothesis[j]=="Ø":
                hypothesis[j]= example[j]

            elif hypothesis[j]!=example[j]:
                hypothesis[j]= "?"

        print(f"After positive example {i+1}: {hypothesis}")
```

```python
print("Final hypothesis:")
print(hypothesis)

print("\nAttribute-wise hypothesis:")
for feature, value in zip(features.columns, hypothesis):
    print(f"{feature}: {value}")
```

```python
mushroom_df= pd.read_csv("mushrooms.csv")

print("Dataset loaded successfully!")
print("Shape:", mushroom_df.shape)
print("\nFirst 5 rows:")
print(mushroom_df.head())
```

```python
print("Column names:")
print(mushroom_df.columns.tolist())

print("\nTarget column: class")
print(mushroom_df["class"].value_counts())

print("\nUnique values in target:")
print(mushroom_df["class"].unique())
```

```python
print("Missing values in each column:")
print(mushroom_df.isnull().sum())

print("\nNumber of '?' values in each column:")
print((mushroom_df=="?").sum())
```

```python
mushroom_clean= mushroom_df.replace("?", pd.NA)
mushroom_clean= mushroom_clean.dropna()

print("Original dataset shape:", mushroom_df.shape)
print("Cleaned dataset shape:", mushroom_clean.shape)

print("\nRemaining missing values:")
print(mushroom_clean.isnull().sum().sum())
```

```python
X_mushroom= mushroom_clean.drop("class", axis=1)
y_mushroom= mushroom_clean["class"]

print("Features shape:", X_mushroom.shape)
print("Target shape:", y_mushroom.shape)

print("\nFeatures:")
print(X_mushroom.head())

print("\nTarget:")
print(y_mushroom.head())
```

```python
hypothesis_mushroom= ["Ø"] * len(X_mushroom.columns)

print("Number of attributes:", len(X_mushroom.columns))
print("\nInitial hypothesis:")
print(hypothesis_mushroom)
```

```python
positive_count= 0

for i in range(len(X_mushroom)):

    if y_mushroom.iloc[i]=="e":
        positive_count+= 1
        example= X_mushroom.iloc[i].values

        for j in range(len(hypothesis_mushroom)):

            if hypothesis_mushroom[j]=="Ø":
                hypothesis_mushroom[j]= example[j]

            elif hypothesis_mushroom[j]!=example[j]:
                hypothesis_mushroom[j]= "?"

print("Number of positive examples processed:", positive_count)
print("\nFinal hypothesis:")
print(hypothesis_mushroom)
```

```python
print("Final Hypothesis - Attribute-wise")
print("=" * 50)

for feature, value in zip(X_mushroom.columns, hypothesis_mushroom):
    print(f"{feature:30} : {value}")
```

```python
print("=" * 60)
print("              FIND-S ALGORITHM RESULTS")
print("=" * 60)

print("\nDataset: Mushroom Classification (Kaggle)")
print("Original samples:", len(mushroom_df))
print("Samples after removing '?':", len(mushroom_clean))
print("Number of attributes:", len(X_mushroom.columns))
print("Positive class: e (Edible)")
print("Negative class: p (Poisonous)")
print("Positive examples processed:", positive_count)

print("\nFinal Hypothesis:")
print(hypothesis_mushroom)

print("\nAttribute-wise Final Hypothesis:")
print("-" * 60)

for feature, value in zip(X_mushroom.columns, hypothesis_mushroom):
    print(f"{feature:30} : {value}")
```
