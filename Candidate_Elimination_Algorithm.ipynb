```python
import pandas as pd
```

```python
mushroom_df = pd.read_csv("/content/mushrooms.csv")

print("Dataset loaded successfully!")
print("Shape:", mushroom_df.shape)

mushroom_df.head()
```

```python
mushroom_clean = mushroom_df.replace("?", pd.NA)

mushroom_clean = mushroom_clean.dropna()

print("Original dataset shape:", mushroom_df.shape)
print("Cleaned dataset shape:", mushroom_clean.shape)

print("Missing values:", mushroom_clean.isnull().sum().sum())
```

```python
X = mushroom_clean.drop("class", axis=1)
y = mushroom_clean["class"]

print("Number of attributes:", len(X.columns))
print("Number of examples:", len(X))

print("\nTarget class distribution:")
print(y.value_counts())

print("\nFeatures:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
num_attributes = len(X.columns)

S = {tuple(["Ø"] * num_attributes)}
G = {tuple(["?"] * num_attributes)}

print("Initial S boundary:")
print(S)

print("\nInitial G boundary:")
print(G)
```

```python
def covers(h, example):
    for i in range(len(h)):
        if h[i] != "?" and h[i] != example[i]:
            return False
    return True


def more_general_or_equal(h1, h2):
    for i in range(len(h1)):
        if h2[i] == "Ø":
            continue

        if h1[i] == "?":
            continue

        if h1[i] != h2[i]:
            return False

    return True
```

```python
def minimal_generalizations(h, example):

    new_h = list(h)

    for i in range(len(h)):

        if new_h[i] == "Ø":
            new_h[i] = example[i]

        elif new_h[i] != example[i]:
            new_h[i] = "?"

    return tuple(new_h)
```

```python
def minimal_specializations(h, example, domains, S):

    specializations = set()

    for i in range(len(h)):

        if h[i] == "?":

            for value in domains[i]:

                if value != example[i]:

                    new_h = list(h)
                    new_h[i] = value
                    new_h = tuple(new_h)

                    if any(more_general_or_equal(new_h, s) for s in S):
                        specializations.add(new_h)

    return specializations
```

```python
domains = [
    sorted(X[column].unique().tolist())
    for column in X.columns
]

print("Attribute domains created successfully!")

for column, values in zip(X.columns, domains):
    print(column, ":", values)
```

```python
for i in range(len(X)):

    example = tuple(X.iloc[i].tolist())
    label = y.iloc[i]

    if label == "e":

        # Process positive example
        G = {g for g in G if covers(g, example)}

        new_S = set()

        for s in S:

            if covers(s, example):
                new_S.add(s)

            else:
                new_h = minimal_generalizations(s, example)

                if any(more_general_or_equal(g, new_h) for g in G):
                    new_S.add(new_h)

        S = new_S

        # Remove more general hypotheses from S
        S = {
            s for s in S
            if not any(
                s != other and more_general_or_equal(s, other)
                for other in S
            )
        }

    elif label == "p":

        # Process negative example
        S = {s for s in S if not covers(s, example)}

        new_G = set()

        for g in G:

            if not covers(g, example):
                new_G.add(g)

            else:
                new_G.update(
                    minimal_specializations(g, example, domains, S)
                )

        G = new_G

        # Remove more specific hypotheses from G
        G = {
            g for g in G
            if not any(
                g != other and more_general_or_equal(other, g)
                for other in G
            )
        }

    # Stop if version space becomes empty
    if not S or not G:
        print("Version space became empty at example:", i + 1)
        break

print("Candidate Elimination completed!")
print("Examples processed:", i + 1)
print("Final S size:", len(S))
print("Final G size:", len(G))
```

```python
print("First 16 training examples:")
print("=" * 60)

for i in range(16):
    print(f"\nExample {i + 1}:")
    print("Class:", y.iloc[i])
    print(X.iloc[i].to_dict())
```

```python
positive_examples = X.iloc[:16][y.iloc[:16] == "e"]
negative_examples = X.iloc[:16][y.iloc[:16] == "p"]

common_values = []

for column in X.columns:
    values = positive_examples[column].unique()

    if len(values) == 1:
        common_values.append((column, values[0]))

print("Attributes shared by all positive examples:")
print(common_values)

print("\nChecking negative examples against these attributes:")

for index, row in negative_examples.iterrows():
    matches = all(row[column] == value for column, value in common_values)
    print(f"Example {index + 1}: {'COVERED' if matches else 'NOT COVERED'}")
```

```python
print("CANDIDATE ELIMINATION - FINAL RESULT")
print("=" * 45)

print("Dataset: Mushroom Classification")
print("Total attributes:", len(X.columns))
print("Examples processed:", i + 1)

print("\nFinal Specific Boundary (S):")
print(S)

print("\nFinal General Boundary (G):")
print(G)

if not S and not G:
    print("\nConclusion:")
    print("The version space is empty.")
    print("No consistent conjunctive hypothesis exists")
    print("for the processed training examples.")
else:
    print("\nConclusion:")
    print("A non-empty version space was obtained.")
```
